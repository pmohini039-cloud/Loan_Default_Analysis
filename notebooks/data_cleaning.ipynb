```python
import pandas as pd
import numpy as np
df=pd.read_csv(r"C:\Users\gujar\OneDrive\Desktop\AVS CODE\Data_Analyst_Portfoio\Loan_Default_Analysis\data\loans_full_schema.csv")
```

```python
missing=pd.DataFrame({
    'Missing Values':df.isnull().sum(),
    'Percentage': (df.isnull().sum()/len(df))*100
})
missing[missing['Missing Values'] > 0].sort_values(
    by='Percentage',
    ascending=False
)
```

```python
df.isnull().sum()
```

```python
cols_to_drop=[
    'annual_income_joint',
    'verification_income_joint',
    'debt_to_income_joint',
    'months_since_90d_late'
]
df.drop(columns=cols_to_drop, inplace=True)
```

```python
df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum().sort_values(ascending=False).head(10)
```

```python
df.drop(columns=['emp_title'], inplace=True)
```

```python
df['emp_length']=df['emp_length'].fillna(
    df['emp_length'].median()
)
```

```python
df['debt_to_income']=df['debt_to_income'].fillna(
    df['debt_to_income'].median()
)
```

```python
df['months_since_last_delinq']=(
    df['months_since_last_delinq']
    .fillna(-1)
)
```

```python
df['months_since_last_credit_inquiry']=(
    df['months_since_last_credit_inquiry']
    .fillna(-1)
)
```

```python
df['num_accounts_120d_past_due']=(
    df['num_accounts_120d_past_due']
    .fillna(0)
)
```

```python
df.isnull().sum().sort_values(ascending=False).head(10)
```

```python
df.duplicated().sum()
```

```python
df['loan_default'] = df['loan_status'].apply(
    lambda x: 1 if x in [
        'Charged Off',
        'Late (31-120 days)',
        'Late (16-30 days)',
        'In Grace Period'
    ] else 0
)
```

```python
df.to_csv(
    "../data/cleaned_loan_data.csv",
    index=False
)
```
